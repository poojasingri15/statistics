###Coin Toss-Sample space

```python
coin_space = ['H','T']
two_coin_space = list(itertools.product(coin_space, coin_space))
print(f"\nTwo-Coin Sample Space : {two_coin_space}")
print(f"\nTotal Outcomes : {len(two_coin_space)}")
```

```python
event_two_heads = [s for s in two_coin_space if s.count("H") == 2]
print(f"\nEvent (2 Heads):{event_two_heads}")
print(f"\nP(2 Heads):{len(event_two_heads)}/{len(two_coin_space)} = {len(event_two_heads)/len(two_coin_space)}")

event_one_heads = [s for s in two_coin_space if s.count("H") == 1]
print(f"\nEvent (1 Heads):{event_one_heads}")
print(f"\nP(1 Heads):{len(event_one_heads)}/{len(two_coin_space)} = {len(event_one_heads)/len(two_coin_space)}")
```

```python
event_atleast_one_heads = [s for s in two_coin_space if s.count("H") >= 1]
print(f"\nEvent (at least one Heads):{event_atleast_one_heads}")
print(f"\nP(at least one Heads):{len(event_atleast_one_heads)}/{len(two_coin_space)} = {len(event_atleast_one_heads)/len(two_coin_space)}")

event_atleast_one_heads = [s for s in two_coin_space if s.count("T") >= 1]
print(f"\nEvent (at least one tails):{event_atleast_one_heads}")
print(f"\nP(at least one tails):{len(event_atleast_one_heads)}/{len(two_coin_space)} = {len(event_atleast_one_heads)/len(two_coin_space)}")
```

##DIce Roll

```python
dice_space = list(range(1,7))
print(f"\nDice Sample Space : {dice_space}")
print(f"\nTotal Outcomes : {len(dice_space)}")

event_even = [x for x in dice_space if x % 2 == 0]
P_A = len(event_even)/len(dice_space)
print(f"\nEvent (Even):{event_even}, P_A = {P_A}")

event_gt3 = [x for x in dice_space if x > 3]
P_B = len(event_gt3)/len(dice_space)
print(f"\nEvent (Greater than 3):{event_gt3}, P_B =  {P_B}")

event_A_and_B = list(set(event_even) & set(event_gt3))
P_A_and_B = len(event_A_and_B)/len(dice_space)
print(f"\nEvent (Even \u2229 Greater Than 3):{event_A_and_B}, P_A_and_B =  {P_A_and_B}")

event_A_OR_B = list(set(event_even) | set(event_gt3))
P_A_or_B = len(event_A_OR_B)/len(dice_space)
print(f"\nEvent (Even \u222A Greater  Than  3):{event_A_OR_B}, P_A_and_B =  {P_A_or_B}")
```

```python
trails_list = [10,100,500,1000,5000,10000]
true_prob = 0.5
print()
print(f"{"Trials":<10} {"Heads":<10} {"Empirical P ":<15} {"Error":<10}")
print("-"*45)
for n in trails_list:
  tossess = np.random.binomial(1, 0.5, n)
  emp_p = tossess.sum() / n
  print(f"{n:<10} {tossess.sum():<10} {emp_p:<15.6f} {abs(emp_p-true_prob):<10.6f}")
```

##Cards

```python
deck = [(rank,suit) for rank in ['2','3','4','5','6','7','8','9','10','A','K','Q','J']
                    for suit in ['Hearts', 'Dimoaands','Clubs','Spades']]

print(deck)

event_king = [c for c in deck if c[0] == 'K']
P_king = len(event_king)/len(deck)
print(f"\nEvent (King):{event_king}, P_king = {P_king}")

event_heart = [c for c in deck if c[1] == 'Hearts']
P_heart = len(event_heart)/len(deck)
print(f"\nEvent (event_heart):{event_heart}, P_king = {P_heart}")

event_king_heart = [c for c in deck if c[0] == 'K' and c[1] == 'Hearts']
P_king_heart = len(event_king_heart)/len(deck)
print(f"\nEvent (event_king_heart):{event_king_heart}, P_king = {P_king_heart}")

p_King_Given_Heart = P_king_heart/P_heart
print(f"p_King_Given_Heart: {p_King_Given_Heart}")
```

BAYE'S THEOREM

```python
print("\n"+"="*60)
print("PART 1D: TOTAL PROBABILITY & BAYES' THEOREM")
print("="*60)

print("\n Medical Diagnosis - bayes' theorem:")

P_Disease=0.01
print(f"\n P(Disease)= {P_Disease:.4f}({P_Disease*100:.1f}%)")

P_No_Disease=1-P_Disease
P_Pos_given_D=0.99
print(f"\n P(Positive|Disease)={P_Pos_given_D:.4f}(Sensitivity)")

P_Neg_given_ND=0.95
P_Pos_given_ND=1-P_Neg_given_ND
print(f"\n P(Positive|Healthly)={P_Pos_given_ND:.4f}(False Positive Rate)")

P_Positive=(P_Pos_given_D * P_Disease + P_Pos_given_ND * P_No_Disease)
print(f"\n P(Positive) by Total Probability ={P_Positive:6f}")

P_D_given_Pos=(P_Pos_given_D*P_Disease)/P_Positive
print(f"\n P(Disease | Positive Test)={P_D_given_Pos:.6f}({P_D_given_Pos*100:.2f}%)")
```
